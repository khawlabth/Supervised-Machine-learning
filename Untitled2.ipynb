### Explication Simple des Variables de Prédiction en Pharmacologie / Biologie (X et Y)

En bio-informatique et modélisation quantitative structure-activité (**QSAR / Modélisation Médicinale**), nous divisons toujours nos données ainsi :

*   **$X$ (Les variables explicatives / Caractéristiques pharmacologiques) :** Ce sont les descripteurs moléculaires, la dose administrée, les propriétés physicochimiques (ex. poids moléculaire, solubilité LogP), ou l'expression de certains gènes.
*   **$Y$ (La variable cible / Effet biologique) :** C'est l'activité biologique, la toxicité, l'inhibition d'une enzyme (ex. $IC_{50}$), ou la réponse clinique du patient que l'on souhaite prédire.

### 🧠 Explication de l'Algorithme d'Apprentissage Automatique : Random Forest (Forêt Aléatoire)

Pour ce projet d'analyse, nous allons utiliser l'algorithme de **Forêt Aléatoire (Random Forest)** pour les raisons suivantes :

1.  **Fonctionnement par Consensus :** Il crée plusieurs arbres de décision indépendants pendant l'entraînement. Chaque arbre vote pour donner une prédiction finale (moyenne pour la régression, majorité pour la classification).
2.  **Robustesse :** Il gère extrêmement bien les relations non linéaires complexes entre vos variables biologiques/pharmacologiques ($X$) et l'effet à prédire ($Y$).
3.  **Évaluation claire :** En comparant les prédictions sur les 20% de données d'évaluation (jeu de test) avec les vraies valeurs, nous pouvons mesurer l'erreur moyenne absolue (MAE) et le score de précision ($R^2$).
